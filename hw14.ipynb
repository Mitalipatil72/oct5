```python
#2. find each name only once
Student_names = ["Mayuri", "Mitali", "vaishu", "Mayuri", "Sakshi","Mitali"]
duplicates = []
for name in Student_names:
    if Student_names.count(name) > 1 and name not in duplicates:
        duplicates.append(name)
        print(duplicates)
        
```

```python
#4 find the names that are not repeated
Student_names = ["Mayuri", "Mitali", "vaishu", "Mayuri", "Sakshi","Mitali"]
for name in set(Student_names):
    if Student_names.count(name) == 1:
        print(name)
```

```python
#5 count the total number of repeated names
Student_names = ["Mayuri", "Mitali", "vaishu", "Mayuri", "Sakshi","Mitali"]
count = 0
for name in set(Student_names):
    if Student_names.count(name) > 1:
        count += 1
print(count)
```

```python
#9 find the most frequently occurring name in the list.
Student_names = ["Mayuri", "Mitali", "vaishu", "Mayuri", "Sakshi","Mitali"]
most_common = max(set(Student_names), key = Student_names.count)
print(most_common)
```

```python
#10 find the least frequently occurring name in the list.
Student_names = ["Mayuri", "Mitali", "vaishu", "Mayuri", "Sakshi","Mitali"]
least_common = min(set(Student_names), key = Student_names.count)
print(least_common)
```
